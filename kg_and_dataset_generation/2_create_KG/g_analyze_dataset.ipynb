# Anlyze the Dataset

First we define some util functions to load the dataset

```python
import gzip
import json
from rdflib import Graph, RDFS, URIRef, BNode, RDF, RDFS, OWL, SKOS
import networkx as nx
import pandas as pd
from collections import defaultdict

def smart_open(filename, mode='rt', encoding='utf-8'):
    if filename.endswith('.gz'):
        return gzip.open(filename, mode=mode, encoding=None if 'b' in mode else encoding)
    else:
        return open(filename, mode=mode, encoding=None if 'b' in mode else encoding)

def convert_tuple_to_rdflib(triplet):
    s, p, o = triplet
    s_rdflib = URIRef(s) if not s.startswith("_:") else BNode(s[2:])
    p_rdflib = URIRef(p) if not p.startswith("_:") else BNode(p[2:])
    o_rdflib = URIRef(o) if not o.startswith("_:") else BNode(o[2:])
    return (s_rdflib, p_rdflib, o_rdflib)

def load_dataset(filename):
    # load the jsonl or json file depending on the file extension (if ending with .gz, use gzip)
    dataset = []
    if filename.endswith('.json.gz') or filename.endswith('.json'):
        with smart_open(filename, mode='rt', encoding='utf-8') as f:
            for line in json.load(f):
                dataset.append(line)
    elif filename.endswith('.jsonl.gz') or filename.endswith('.jsonl'):
        with smart_open(filename, mode='rt', encoding='utf-8') as f:
            for line in f:
                dataset.append(json.loads(line))
    else:
        raise ValueError(f"Unsupported file format: {filename}")
    
    for example in dataset:
        # convert all triplets to tuples
        example["triplets"] = [convert_tuple_to_rdflib(triplet) for triplet in example["triplets"]]
    return dataset
    
def load_graph(filename):
    g = Graph()
    with smart_open(filename, mode='rb') as f:
        g.parse(file=f, format="nt")
    return g
```

```python
test_dataset = load_dataset("test_dataset.jsonl.gz")
val_dataset = load_dataset("val_dataset.jsonl.gz")
train_dataset = load_dataset("train_dataset.jsonl.gz")
```

```python
def extract_union_members(union_node, graph):
    """Return classes inside owl:unionOf lists."""
    members = set()
    for collection in graph.objects(union_node, OWL.unionOf):
        for item in graph.items(collection):
            members.add(item)
    return members

def get_domain_or_range(prop, predicate, graph):
    """Retrieve domain or range, including unionOf constructs."""
    results = set()
    for obj in graph.objects(prop, predicate):
        # Case 1: direct class
        if (obj, OWL.unionOf, None) in graph:
            results |= extract_union_members(obj, graph)
        else:
            results.add(obj)

    # add all subclasses of the classes found
    all_results = set(results)
    for cls in results:
        if cls in subclass_graph:
            subclasses = nx.ancestors(subclass_graph, URIRef(cls))
            all_results |= subclasses
    results = all_results

    return results

def get_all_properties_domain_range(graph):
    prop_to_domain = defaultdict(set)
    prop_to_range = defaultdict(set)
    for property_type in [OWL.DatatypeProperty, OWL.ObjectProperty, RDF.Property]:
        for prop in graph.subjects(RDF.type, property_type):
            prop_to_domain[prop] |= get_domain_or_range(prop, RDFS.domain, graph)
            prop_to_range[prop] |= get_domain_or_range(prop, RDFS.range, graph)
    return prop_to_domain, prop_to_range

prop_to_domain, prop_to_range = get_all_properties_domain_range(prop_graph)
print(list(prop_to_domain.items())[:5])
print(list(prop_to_range.items())[:5])

k = 0
for prop_key, domain_ranges in prop_to_domain.items():
    if len(domain_ranges) > 1:
        k += 1
        print(prop_key, domain_ranges)
        if k > 5: 
            break
```

```python
def general_anaylsis(dataset):
    list_of_number_of_triples = []
    missing_subject, missing_predicate, missing_object = 0, 0, 0
    for example in dataset:
        triple_count = 0
        for s, p, o in example["triplets"]:
            triple_count += 1

            if type(s) == BNode:
                missing_subject += 1
            if type(p) == BNode:
                missing_predicate += 1
            if type(o) == BNode:
                missing_object += 1

        list_of_number_of_triples.append(triple_count)
        
    print(pd.Series(list_of_number_of_triples).describe())
    # print sum of all triples
    print("All triples:", sum(list_of_number_of_triples))


    print("Missing subjects:", missing_subject)
    print("Missing predicates:", missing_predicate)
    print("Missing objects:", missing_object)

print("test data")
general_anaylsis(test_dataset)

print("val data")
general_anaylsis(val_dataset)

print("train data")
general_anaylsis(train_dataset)
```

```python
def get_label_set(graph, entity):
    labels = set()
    for _, _, label in graph.triples((entity, RDFS.label, None)):
        labels.add(str(label))
    if not labels:
        for _, _, label in graph.triples((entity, SKOS.altLabel, None)):
            labels.add(str(label))
    return labels

def get_label_set(graph, entity):
    for _, _, label in graph.triples((entity, RDFS.label, None)):
        return str(label)
    for _, _, label in graph.triples((entity, SKOS.altLabel, None)):
        return str(label)
    return None

#analyze how often the domain and range is correct 
def analyze_dataset(dataset):
    no_domain_available = 0
    no_range_available = 0
    domain_matches = 0
    range_matches = 0
    domain_incorrect = 0
    range_incorrect = 0
    for example in dataset:
        should_break = False
        for s, p, o in example["triplets"]:
            # our own properties only
            if type(p) == URIRef and str(p).startswith("http://www.myiedata.org/") and \
                type(s) == URIRef and type(o) == URIRef:
                 #type(s) == URIRef and str(s).startswith("http://www.myiedata.org/") and \
                 #   type(o) == URIRef and str(o).startswith("http://www.myiedata.org/"):
                expected_domains = prop_to_domain.get(p, set())
                if not expected_domains:
                    no_domain_available += 1
                else:
                    s_types = inst_to_type.get(s, set())
                    if s_types.intersection(expected_domains):
                        domain_matches += 1
                    else:
                        domain_incorrect += 1
                        #print("Domain incorrect:", get_one_label(s), get_one_label(p), get_one_label(o), s_types, expected_domains)
                        
                        #print("Domain incorrect:")
                        #print(f"     subject ({s}): {get_label_set(entities_graph, s)}")
                        #print(f"     predicate ({p}): {get_label_set(prop_graph, p)}")
                        #print(f"     object ({o}): {get_label_set(entities_graph, o)}")
                        #print("     Subject types:")
                        #for st in s_types:
                        #    print(f"          {st} : {get_label_set(classes_graph, st)}")
                        #print("     Expected domain:")
                        #for ed in expected_domains:
                        #    print(f"          {ed} : {get_label_set(classes_graph, ed)}")
                        #should_break = True
 
                
                expected_ranges = prop_to_range.get(p, set())
                if not expected_ranges:
                    no_range_available += 1
                else:
                    o_types = inst_to_type.get(o, set())
                    if o_types.intersection(expected_ranges):
                        range_matches += 1
                    else:
                        range_incorrect += 1
        if should_break:
            break

    print("No domain available:", no_domain_available)
    print("Domain matches:", domain_matches)
    print("Domain incorrect:", domain_incorrect)
    print("No range available:", no_range_available)
    print("Range matches:", range_matches)
    print("Range incorrect:", range_incorrect)

print("Analyzing test dataset")
analyze_dataset(val_dataset)

print("===")
print("Analyzing val dataset")
analyze_dataset(test_dataset)

print("===")
print("Analyzing train dataset")
analyze_dataset(train_dataset)
```
